We will use a tiny customer-support classifier with four input features:

- `refund`
- `broken`
- `thank_you`
- `general_question`

The model produces two scores:

- `urgent`
- `normal`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F

np.set_printoptions(precision=3, suppress=True)
torch.set_printoptions(precision=3, sci_mode=False)

feature_names = ["refund", "broken", "thank_you", "general_question"]
class_names = ["urgent", "normal"]
```

I want a refund, I got a broken phone.

```python
x = np.array([1.0, 1.0, 0.0, 0.0])

pd.DataFrame({
    "feature": feature_names,
    "value": x
})
```

```python
W_base = np.array([
    [0.10, 0.20, -0.10, 0.10],  # urgent weights
    [0.50, 0.40,  0.30, 0.50]   # normal weights
])

base_weight_table = pd.DataFrame(
    W_base,
    index=class_names,
    columns=feature_names
)

base_weight_table
```

We've 4 features, f1, f2, f3 and f4

And these are corresponding weights for them, 16, 2, -8, 4




```python
base_scores = W_base @ x

base_score_table = pd.DataFrame({
    "class": class_names,
    "score": base_scores
})

base_prediction = class_names[np.argmax(base_scores)]

print(base_score_table.to_string(index=False))
print(f"\nBase prediction: {base_prediction}")
```

```python
delta_W = np.array([
    [ 0.60,  0.50, 0.00, 0.00],
    [-0.18, -0.15, 0.00, 0.00]
])

delta_weight_table = pd.DataFrame(
    delta_W,
    index=class_names,
    columns=feature_names
)

delta_weight_table
```

```python
W_effective = W_base + delta_W
```

```python
adapted_scores = W_effective @ x
adapted_prediction = class_names[np.argmax(adapted_scores)]

score_comparison = pd.DataFrame({
    "class": class_names,
    "base score": base_scores,
    "adapted score": adapted_scores
})

print(score_comparison.to_string(index=False))
print(f"\nBase prediction: {base_prediction}")
print(f"Adapted prediction: {adapted_prediction}")
```
